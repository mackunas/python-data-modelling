```python
import numpy as np
import random
x=np.arange(0,100)
y=np.zeros(len(x))
z=np.zeros(len(x))
for i in range(len(x)):
    y[i] = 2.0 * x[i] + 50 * random.random()
    z[i] = 250 * random.random() - 2.0 * x[i]
```
