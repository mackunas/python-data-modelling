## Classes and Objects

A class is a description that defines the data structure of objects of a certain type and the rules for managing them.

A class is a data type defined by the programmer, which allows creating objects of this data type.

An object is a unique set of data corresponding to the structure described in that object's class. An object can be called an instance of the class.

A class consists of a declaration (the `class` statement), the class name, and the class body, which consists of attributes and methods:

```
class Class_name:
    [ '''documentation line''' ]
    statements
```

Minimal class declaration:

````python
class C:
    pass
````

An object, or class instance, is created by calling the class as a function.

````python
c_obj = C() # object c_obj created
````

The method `isinstance(O, K)` checks whether object O is an object of class K or of any of its subclasses.

````python
class C:
    pass

c_obj = C()

print(isinstance(c_obj, C))
````

````python
class C:
    pass

c_obj = C()

print(c_obj)
````

The class documentation line can be retrieved using the `__doc__` method:

`Class_name.__doc__`

````python
class Rectangle:
    '''Class describes the properties of a rectangle'''

print(Rectangle.__doc__)
````

## Constructor

A class can have a method `__init__()`, which is called the class constructor or initialization method.

The method `__init__()` must have the parameter `self`.

The method `__init__()` is called automatically every time the class is used to create a new object:

````python
class Test:
    def __init__(self):
        print('Constructor execution...')

t1 = Test() # object t1 created
t2 = Test() # object t2 created
````

Since the `__init__()` method initializes attributes, it must not return a value; otherwise a `TypeError` occurs.

Object attributes are accessed using the dot "." operator

````python
class LT_miestai:
    def __init__(self, m, atst):
        self.miestas = m # object attribute initialization
        self.atstumas = atst # object attribute initialization

m1 = LT_miestai('Akmenė', 274) # object m1

print('Miestas -', m1.miestas)
print('Atstumas iki sostinės -', m1.atstumas,'\n')

m2 = LT_miestai('Birštonas', 91) # object m2

print('Miestas -', m2.miestas)
print('Atstumas iki sostinės -', m2.atstumas)
````

````python
class C:
    def __init__(self, name, lno):
        self.name = name
        self.lno = lno

cat = C('Cat',4) # object cat
bird = C('Bird',2) # object bird

print('{} has {} legs'.format(cat.name, cat.lno))
print('{} has {} legs'.format(bird.name, bird.lno))
````

````python
class C:
    def __init__(self, *t): # *t - any number of positional parameters
        self.tpl = t

t1 = C(1,2,3,4,5) # object t1
t2 = C(200) # object t2
t3 = C() # object t3
print(t1.tpl)
print(t2.tpl)
print(t3.tpl)
````

````python
class C:
    def __init__(self,**d): # **d - any number of keyword parameters
        self.dct = d

d1 = C(a=1,b=2,c=3) # object d1
d2 = C(A=100) # object d2
d3 = C() # object d3

print(d1.dct)
print(d2.dct)
print(d3.dct)
````

If the class does not have an `__init__()` method, the object is created without its characteristic attributes.

## Variables

### Instance variables (object-level variables)

Ways to declare instance variables:
- in the constructor, using the keyword `self`;
- in an instance method (discussed later), using the keyword `self`;
- outside the class, using a reference to the object (instance).

The built-in class attribute `__dict__` allows viewing the instance variables. They are displayed as a dictionary.

To perform operations with instance variables, an instance of the class must be created.

````python
class Test:
    def __init__(self, a): # constructor
        self.a = a
        self.b = 20
        
    def m1(self, d): # instance method
        self.c = 30
        self.d = d

t = Test(10) # object t created
print('t.__dict__:',t.__dict__)

t.m1(40) # instance method m1 called
print('t.__dict__:',t.__dict__)

t.e = 50 # instance attribute e created
print('t.__dict__:',t.__dict__,'\n')

t2 = Test(100) # object t2 created
print('t2.__dict__:',t2.__dict__)
````

Operations with instance variables can be performed inside a class method and outside the class, using a reference to the object:

````python
class Test:
    def __init__(self):
        self.a = 10
        self.b = 20
        
    def info(self):
        print(self.a)

t = Test() # instance t created
t.info() # instance method info called
print(t.b)
````

The value of an instance variable can be changed both inside the class and outside the class:

````python
class Test:
    def __init__(self):
        self.a = 10
        self.b = 20
        
    def m(self):
        self.a = 1000 # changing the value of variable a

t = Test() # object t
print(t.__dict__)

t.m() # instance method m called
print(t.__dict__)

t.a = 100000 # changing the value of variable a
print(t.__dict__,'\n')

t2 = Test() # object t2
print(t2.__dict__)
````

An instance variable can be deleted both inside the class and outside the class:

````python
class Test:
    def __init__(self):
        self.a = 10
        self.b = 20
        self.c = 30
        
    def m(self):
        del self.a # variable a deleted

t = Test() # object t
print(t.__dict__)

t.m() # instance method m called
print(t.__dict__)

del t.b # variable b deleted
print(t.__dict__)
````

### Static variables (class-level variables)

Static variables are shared by all objects of the class.

To perform operations with a static variable, it is not necessary to create an instance of the class.

Ways to declare static variables:
- directly in the class, outside the constructor or other methods;
- in the constructor, using the class name;
- in an instance method (discussed later), using the class name;
- in a class method (discussed later), using `cls` or the class name;
- in a static method (discussed later), using the class name.

````python
# Outside the constructor or other methods
class Test:
    a = 'Static variable'
    
    def __init__(self):
        self.b = 10

print(Test.a)
````

````python
# In the constructor, using the class name
class Test:
    def __init__(self):
        self.b = 10
        Test.a = 'Static variable'

t = Test()
print(Test.a)
````

````python
# In an instance method, using the class name
class Test:
    def i_method(self):
        Test.a = 'Static variable'

t = Test()
t.i_method()
print(t.a)
````

````python
# In a class method, using the class name
class Test:

    @classmethod
    def c_method(cls):
        cls.a = 'Static variable'

t = Test()
t.c_method()
print(t.a)
````

````python
# In a static method, using the class name
class Test:
    @staticmethod
    def s_method():
        Test.a = 'Static variable'

t = Test()
t.s_method()
print(Test.a)
````

Operations with static variables can be performed in the class:
- in the constructor, using `self` or the class name (the class name is recommended);
- in an instance method, using `self` or the class name (the class name is recommended);
- in a class method, using `cls` or the class name;
- in a static method, using the class name;

````python
class Test:
    a = 'Static variable' # static variable
    
    def __init__(self): # constructor
        print('Constructor:',Test.a)
        
    def i_method(self): # instance method
        print('Instance method:',Test.a)
        
    @classmethod
    def c_method(cls): # class method
        print('Class method:',cls.a)
        
    @staticmethod
    def s_method(): # static method
        print('Static method:',Test.a)

t = Test()

t.i_method()
t.c_method()
t.s_method()

print('Object reference:',t.a)
````

The value of a static variable can be changed in a class method using `cls`, or anywhere else using the class name (`self` and an object reference cannot be used!):

````python
class Test:
    a = 10 # static variable
    
    def __init__(self):
        Test.a = 20 # changing the value of variable a
        
    def i_method(self):
        Test.a = 30 # changing the value of variable a
        
    @classmethod
    def c_method(cls):
        cls.a = 40 # changing the value of variable a
        
    @staticmethod
    def s_method():
        Test.a = 50 # changing the value of variable a

print(Test.a)

t = Test() # constructor called
print(Test.a)

t.i_method() # instance method called
print(Test.a)

t.c_method() # class method called
print(Test.a)

t.s_method() # static method called
print(Test.a)
````

A static variable can also be accessed through a class instance — everything works fine as long as the variable's value is not changed.

If the value of a static attribute is changed through a class instance, a variable with the same name as the static one appears in the instance, and access to the static variable is lost:

````python
class Rectangle:
    color = 'grey' # static variable

r1 = Rectangle()
r2 = Rectangle()

print('r1.color -', r1.color)
print('r2.color -', r2.color,'\n')

print('Rectangle.color -', Rectangle.color)

r1.color = 'red' # instance variable color created

print('Rectangle.color -', Rectangle.color,'\n')

print('r1.color -', r1.color)
print('r2.color -', r2.color)
````

A static variable can only be deleted using `cls` or the class name.

### Local variables (method-level variables)

Local variables are variables declared directly inside a method. Local variables are created during the execution of the method and destroyed when the method's execution ends. Local method variables cannot be accessed outside the method.

````python
class Test:
    def m1(self):
        a = 1000
        print(a)
        
    def m2(self):
        b = 5000
        print(b)

t = Test()
t.m1()
t.m2()
````

````python
class Test:
    def m1(self):
        a = 1000

    def m2(self):
        b = 5000
        print(b)
        print(a) # printing the local variable a of method m1

t = Test()
t.m2()
````

## Methods

A method is a function inside a class.

Methods can be:
- instance methods;
- class methods;
- static methods.

### Instance method

An instance method is used to perform operations with instance variables.

The first argument of the method is `self`.

To call an instance method, an instance must be created.

When calling the method, the `self` argument does not need to be included in the argument list.

````python
class LT_miestai:
    def __init__(self, m, atst):
        self.m = m
        self.atst = atst
        
    def info(self):
        print('Miestas -', self.m)
        print('Atstumas iki sostinės -', self.atst)

m1 = LT_miestai('Akmenė', 274)
m2 = LT_miestai('Birštonas', 91)

m1.info()
print()

m2.info()
````

````python
class LT_miestai:
    def mst(self, miestas):
        self.miestas = miestas
        
    def atst(self, atstumas):
        self.atstumas = atstumas
        
    def info_mst(self):
        return self.miestas # returns the value of variable miestas
        
    def info_atst(self):
        return self.atstumas # returns the value of variable atstumas

n = int(input('Miestų skaičius: '))
Miestai = [] # empty list

# n objects are created in the loop. All objects are stored in the list Miestai
for i in range (n):
    m = LT_miestai() # object m
    
    miestas = input('Įveskite miesto pavadinimą: ')
    atstumas = int(input('Įveskite atstumą iki sostinės: '))
    
    m.mst(miestas) # method mst called
    m.atst(atstumas) # method atst called

    Miestai.append(m)
    print()

print(Miestai) # Miestai - list of objects

for m in Miestai:
    print()
    print('Miestas:', m.info_mst())
    print('Atstumas iki sostinės:', m.info_atst())
````

### Class method

Class methods are bound to the class itself, not to its instance.

Class methods can change a class variable, and this will affect all objects of this class; however, these methods cannot change instance variables.

A mandatory `@classmethod` decorator is written before a class method; the first argument of the method is `cls`.

A class method can be called not only via an object but also via the class.

````python
class Test:
    a = 10
    
    @classmethod
    def m1(cls):
        cls.a = 1000 # changing the value of static variable a

t1 = Test()
t2 = Test()

print('t1.a:', t1.a)
print('t2.a:', t2.a)

Test.m1() # or t1.m1, or t2.m1
print()

print('t1.a:', t1.a)
print('t2.a:', t2.a)
````

````python
class Test:
    nr_obj = 0 # static variable

    def __init__(self):
        Test.nr_obj += 1 # changing the value of static variable nr_obj
        
    @classmethod
    def info(cls):
        print('Klasės Test objektų skaičius -',cls.nr_obj) # printing the value of nr_obj

t1 = Test()
t2 = Test() 

Test.info()

t3 = Test()
t4 = Test()

t1.info()
````

### Static method

A static method is a method that can be called not only via an object but also via the class.

A `@staticmethod` decorator is written before a static method.

This method does not require a specific first argument.

````python
class Test:

    @staticmethod
    def m2(a,b):
        return a*a, b*b

print(Test.m2(2, 4))
````

````python
class Test:
    a = 10 # static variable

    @staticmethod
    def m1():
        Test.a = 10000 # changing the value of static variable a

    @staticmethod
    def m2():
        print('a =',Test.a) # printing variable a

print(Test.a)
Test.m1()
Test.m2()
````

Accessing members of one class from inside another class:

````python
class Employee:
    def __init__(self,name,salary=2000):
        self.name = name
        self.salary = salary
        
    def info(self):
        print('Employee name :',self.name)
        print('Employee salary:',self.salary,'\n')

class Sal:
    @staticmethod
    def modify(empl,n):
        empl.salary = empl.salary + n
        empl.info()

e1 = Employee('Smith') # object e1
e1.info()

e2 = Employee('Williams',2500) # object e2
e2.info()

Sal.modify(e1, 800) # object e1 passed to method modify as an argument
````
