### The `eval()` Function

eval() evaluates a string as a Python expression. Use it carefully because it can execute arbitrary code.

````python
b = bool(input('Enter True or False: '))
print('b =', b)
````

When converting to the Boolean (`bool`) type, all strings except empty strings (`''`) are considered `True`. Therefore, in this case, the `eval()` function is used for type conversion:

````python
b = eval(input(' Enter True or False: '))
print('b =', b)
````

`eval()` isn't limited to `True`/`False` — it evaluates any Python expression the user types in. For example, a full arithmetic expression is evaluated, not just read as text:

````python
b = eval(input('Enter Expression: '))
print('b =', b)
````

## Mutable and Immutable Objects

A **mutable object** is an object that can change its value while keeping the same `id()`.

An **immutable object** is an object with a fixed value.

If we want to change the value of an immutable object, a new object must be created.

### The `id()` Function

Every Python object is stored in a specific memory location.

The `id()` function returns the memory location of a given object. The address is returned as a base-10 number.

This function can be used to check whether objects refer to the same memory location, meaning they refer to the same data in memory.

### The `is` Keyword

`is` is a keyword that compares two objects and returns `True` if both objects have the same memory address.

`is not` returns `False` if both objects have the same memory address.

### Immutable Objects

- Numbers (`int`, `float`, `decimal`, `boolean`, `complex`)
- Strings
- Tuples
- Frozen sets

Reassigning a variable to a new value creates a new object, so `id(a)` changes:

````python
a = 10
print('a =', a, ' id(a)=', id(a))
a = 20
print('a =', a, ' id(a)=', id(a))
````

When several variables are assigned from the same object, they all share the same `id` and `is` returns `True`:

````python
a = 1000
b = a
c = a
print('id(a)=', id(a))
print('id(b)=', id(b))
print('id(c)=', id(c))
````

````python
print('a is b ->', a is b)
print('a is c ->', a is c)
print('b is c ->', b is c)
````

Small integers are cached by Python, so two variables holding the same small value may be `is`-identical. But once a variable is reassigned to a different value, `is` no longer holds:

````python
a = 10
b = a
print('a is b ->', a is b)
a = 1000
print('a is b ->', a is b)
````

Mutable objects include:

- Lists
- Sets
- Dictionaries

A **list** is a collection of values of any data type, separated by commas and enclosed in square brackets.

Lists are mutable: changing an element in place does not change the list's `id`:

````python
my_list = [10, 20, 30, 40]
print(id(my_list))
my_list[2] = 3000
print(id(my_list))
print(my_list)
````

Assigning a list to another variable does not copy it — both names refer to the same object, so modifying one affects the other:

````python
list1 = [10, 20, 30, 40]
list2 = list1
print('list1 is list2 ->', list1 is list2)
list1[0] = 100.52
print('list1 is list2 ->', list1 is list2)
print('list1 =', list1)
print('list2 =', list2)
````

Two different lists that happen to hold equal values are still two distinct objects:

````python
list1 = [10, 20, 30, 40]
list2 = [10, 20, 30, 40]
print('id(list1)=', id(list1))
print('id(list2)=', id(list2))
````

Two separate empty lists are equal in value (`==`) but are not the same object (`is`):

````python
x = []
y = []
print('x == y ->', x == y)
print('x is y ->', x is y)
print('x is not y ->', x is not y)
````

Tuples are immutable: trying to change an element raises a `TypeError`:

````python
t = (1, 2, 3) # tuple
try:
    t[0] = 99
except TypeError as e:
    print('TypeError:', e)
````

Sets and dictionaries are mutable too, so items can be added after creation. A `frozenset` is the immutable counterpart of a `set`:

````python
s = {1, 2, 2, 3} # set
s.add(4)
print('set:', s)
d = {'a': 1, 'b': 2} # dictionary
d['c'] = 3
print('dict:', d)
fs = frozenset([1, 2, 2, 3])
print('frozenset:', fs)
````

## Input and Output

### Data Input

Data is entered from the keyboard using the `input()` function.

All data read from the keyboard is treated as strings (`<class 'str'>`).

To avoid calculation errors, the input data type must be converted during input or before performing calculations.

Without conversion, `input()` always returns a string, so `+` concatenates the text instead of adding numbers:

````python
a = (input('First number: '))
b = (input('Second number: '))
print('a + b =', int(a)+int(b))
````

To add the numbers directly (without converting inside `print`), convert the input to `int` right away:

````python
a = int(input('First number: '))
b = int(input('Second number: '))
print('a + b =', a+b)
````

### Data Output

The `print()` function displays output on the screen.

Special characters:
- `\n` creates a new line.
- `\t` adds a tab space.

### The `sep` Attribute

The `sep` argument changes the separator between multiple values.

````python
x, y = 2.5, 19
print(x, y, x*y)
print(x, y, x*y, sep='-----')
print(x, y, x*y, sep='      ')
````

### The `end` Attribute

The `end` argument changes what is printed at the end (default is a new line).

````python
x, y = 2.5, 19
print(x, end='*****')
print(y, end='\n\n\n')
print(x*y)
````

## Formatted output
### The String Formatting Method `.format()`

The `format()` method inserts values into a string using `{}` placeholders.

Ways to use it:
- `{}` uses values in order.
- `{0}`, `{1}`, etc. use positions.
- `{name}` uses named arguments.

````python
first_name = 'John'
last_name = 'Smith'
salary = 2500
print('Hello {} {}. Your salary is {}'.format(first_name, last_name, salary))
print('Hello {0} {1}. Your salary is {2}'.format(first_name, last_name, salary))
print('Hello {1} {0}. Your salary is {2}'.format(first_name, last_name, salary))
print('Hello {a} {b}. Your salary is {c}'.format(a=first_name, b=last_name, c=salary))
print('Hello {} {}. Your salary is {}'.format('John', 'Smith', 2500))
````

### Output Formats

- `:d` – integer (`int`)
- `:f` – floating-point number (`float`)
- `:e` – exponential notation
- `:s` – string (`str`)

````python
n = 458.541315
try:
    print('{:d}'.format(n))  # wrong: n is float
except ValueError as e:
    print('ValueError:', e)
print('{:d}'.format(int(n)))
print('{:f}'.format(n))
print('{0:15f}'.format(n))
print('{0:15e}'.format(n))
print('{0:15.3f}'.format(n))
````

### Alignment

- `<` – align to the left
- `>` – align to the right
- `^` – align to the center

````python
print('{0:15f}{1}'.format(n, '|'))
print('{0:<15f}{1}'.format(n, '|'))
print('{0:>15f}{1}'.format(n, '|'))
print('{0:^15.3f}{1}'.format(n, '|'))
print('{:s}'.format('hello'))
````

## Formatting Using f-strings

This formatting method has been available since Python 3.6.

f-strings allow variables and expressions to be inserted directly into strings using {}.

To format a string, the letter `f` is placed before the string.

````python
x = 356.089775672
print(f'Value of x: {x}')
print(f'{x = }')
````

They also support formatting options:

Set the display width.
Control the number of decimal places.

````python
width = 10
precision = 4
y = 12.3456789
print(f'Value of y: {y:{width}.{precision}}')
````

- **Width** specifies the minimum amount of space used to display a value.
- **Precision** specifies the number of significant digits displayed for a number.

A simpler way to fix the number of decimal places is to write the precision directly in the format spec:

````python
y = 12.3456789
print(f'Value of y: {y:.2f}')
````

## Standard Python Modules

A standard Python installation includes a library of standard modules.

A **module** is a file with the `.py` extension that contains a collection of functions, classes, variables, constants, and other Python objects used to solve specific tasks.

The contents of a module can be accessed and used by any other Python program.

You can display a list of the standard modules by running the following command:

```python
help("modules")
```

````python
# uncomment to see a list of the standard modules
# help('modules')
````

A module is imported using the following command:

```python
import module_name
```

To call a function from a module, write the module name followed by a dot (`.`) and the function name:

```python
module_name.function_name()
```

````python
import math

print(math.factorial(5))      # 120
x = math.pi
print(math.cos(x))            # -1.0
````

Multiple modules can be imported at the same time:

```python
import module_name_1, module_name_2, ..., module_name_n
```

Here `math` computes an angle in degrees:

````python
import math, random

print(math.degrees(math.asin(1)))  # 90.0
````

...and `random`, imported alongside it, generates a random integer:

````python
print(random.randint(-5, 5))  # random integer in [-5, 5]
````

A module can also be imported using a different name (an **alias**):

```python
import module_name as new_module_name
```

````python
import math as m
print(m.pi)
print(m.e)
````

The `from` statement allows you to import specific functions from a module:

```python
from module_name import function_name[, ..., function_name_n]
```

After importing the functions this way, you can call them directly without writing the module name:

```python
function_name()
```

````python
from math import sin, pi
print(sin(pi/4))  # 0.7071067811865476
````
