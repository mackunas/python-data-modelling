## Programming style (PEP 8)
Most programming languages can be written (or, more accurately, formatted) in different styles. Some styles are easier to read than others. Writing code that is easy for others to understand is a good practice, and using a consistent coding style helps make programs more readable.

Most Python projects follow the style guidelines defined in **PEP 8**, which promote clear and readable code. Some of the most important recommendations are:

- Use **4 spaces** for indentation, and do not use tabs.
- Use blank lines to separate functions, classes, and larger blocks of code.
- Whenever possible, write comments on a separate line.
- Use **docstrings** to document functions, classes, and modules.
- Use spaces around operators and after commas, but not inside parentheses:

```python
a = f(1, 2) + g(3, 4)
```

- Use consistent naming conventions:
  - **Classes:** `CamelCase`
  - **Functions and methods:** `lower_case_with_underscores`
  - Always use `self` as the first parameter of instance methods.

## Variables and data types


### Variable

Python is an **object-oriented programming language**, where all elements (numbers, strings, functions, etc.) are treated as objects.

A **variable name** is a reference to an object.

A value is assigned using the `=` symbol:

```python
variable_name = value
```

````python
a = 100
b = 'one hundred'
print(a, b)
````

Python allows you to assign the same or different values to multiple variables at the same time:

````python
a = b = c = 10
print(a, b, c)
````

````python
x, y, z = 1, 2, 'five'
print(x, y, z)
````

If an assigned value is too long, it can be continued on the next line using the `\` symbol:

````python
line = 'the quick brown fox jumps \
over a lazy dog'
print(line)
````

If brackets `[ ]`, `{ }`, or `( )` are used, there is no need to use the line continuation symbol `\`:

````python
A = [1, 2, 3, # list
     4, 5, 6,
     7, 8, 9]
print(A) # output
````

The `#` symbol indicates that everything written after it is a comment.

A comment can consist of only one line. If a comment needs to span multiple lines, the `#` symbol must be written at the beginning of each line.

### Variable Naming Rules

Variable names in Python must follow these rules:

- They may contain letters, numbers, and underscores (`_`).
- They cannot begin with a number.
- They cannot contain spaces.
- Python is **case-sensitive**, so uppercase and lowercase letters are treated as different.
- A variable name cannot be the same as a Python keyword.

You can display the list of Python keywords using the `keyword` module:

````python
import keyword
print(keyword.kwlist)
````

````python
print(keyword.iskeyword('if')) # Checks if 'if' is a Python keyword
````

````python
print(keyword.iskeyword('iff'))
````

## Core data types
Python is a **dynamically typed programming language**. This means that variables do not have a fixed type; instead, the type belongs to the value assigned to the variable.

The type of a variable can be determined using the `type` command:

```python
type(variable_name)
```

### Integers

Whole numbers, e.g. `10`, `-5`

````python
print(2 + 3)
x = 10
print(type(x), x)
````

### Float

Numbers with decimal points, e.g. `2.1`, `-6.3`

````python
a = 2.1
b = 3 * a
print(b, type(b))
print(-6.3e2)
````

````python
a = 10.2e-4
print(a, type(a))
````

### Complex Numbers

Complex numbers consist of a real part (`.real`) and an imaginary part (`.imag`):

````python
c = 5.1 + 0.1j
print(c.real, c.imag, type(c))
````

````python
d = 2 - 1.4j
print(c - d)
````

### Binary Integers (`bin`)

Binary integers are base-2 whole numbers. They start with `0b` or `0B` (zero followed by the letter `b` or `B`).

````python
print(0b11011)   # 27
print(bin(15))
````

### Octal Integers (`oct`)

Octal integers are base-8 whole numbers. They start with `0o` or `0O` (zero followed by the letter `o` or `O`).

````python
print(0o45601)   # 19329
print(oct(2020))
````

### Hexadecimal Integers (`hex`)

Hexadecimal integers are base-16 whole numbers. They start with `0x` or `0X` (zero followed by the letter `x` or `X`).

````python
print(0x15ab3)   # 88755
print(hex(1027297))
````

### Booleans

Python uses the **`bool`** data type to represent logical values:

- `True` — represents a condition that is correct
- `False` — represents a condition that is incorrect

Comparison operators return Boolean values:
- `>` greater than
- `<` less than
- `==` equal to
- `!=` not equal to

````python
print(1 > 3)
a, b = 2, 3
print(a != b)
````

````python
boo = (3 > 1)
print(boo, type(boo))
````

Boolean values can also behave like numbers:
- `True` is treated as `1`
- `False` is treated as `0`

````python
print(True + True)
print(False - True)
print(True * False)
````

Dividing by `False` behaves like dividing by `0`, so it raises an error:

````python
try:
    print(True / False)
except ZeroDivisionError as e:
    print('ZeroDivisionError:', e)
````

### Strings

A **string (`str`)** stores text in Python. Strings are assigned using single or double quotation marks:

````python
s1 = 'string value'
s2 = "string value"
print(type(s1), type(s2))
````

An empty string can be created.

The built-in `len()` function returns the length of a string.

````python
print(len(''))   # empty string -> 0
print(len(' '))  # space -> 1
````

If a string contains a single quotation mark, it can be enclosed in double quotation marks, and vice versa. Alternatively, the escape character `\` can be used.

Triple single or double quotation marks can also be used.
```

Using double quotes to wrap a string that itself contains single quotes:

````python
s = '"Yes," he said'
print(s)
````

Using the escape character `\` to include quotation marks inside a string of the same type:

````python
s = '\"Isn\'t,\" she said'
print(s)
````

Using triple quotes so no escaping is needed at all:

````python
s = '''"Isn't," she said'''
print(s)
````

Triple quotation marks allow you to create a string with the desired format.

````python
multi = '''Hello
World'''
print(multi)
````

Strings can be joined using the `+` operator and repeated using the `*` operator:

````python
print('Hello' + ' ' + 'world')
print('Hello ' * 3 + 'world')
print(5 * 'Hello ')
````

Strings written next to each other are automatically joined:

````python
print('Hello' ' ' 'world')
````

## Data Type conversion
A data type can be converted using the functions `int()`, `float()`, `bool()`, `complex()`, and `str()`.

### `int()`

Converting a `float` to `int` truncates the decimal part:

````python
print(int(10.999))  # float > int
````

A `complex` number cannot be converted to `int`:

````python
try:
    print(int(10 + 5j)) # complex > int
except TypeError as e:
    print('TypeError:', e)
````

`bool` converts to `0` or `1`, and a numeric string converts directly:

````python
print(int(True), int(False)) # bool > int
print(int('20')) # str > int
````

Strings that aren't plain base-10 integers raise a `ValueError`:

````python
try:
    print(int('10.5'))
except ValueError as e:
    print('ValueError for 10.5:', e)

try:
    print(int('0b1111'))
except ValueError as e:
    print('ValueError for 0b1111:', e)
````

### `float()`

`int` values, including those written in binary, octal, or hex, all convert to `float`:

````python
print(float(100)) #int > float
print(float(0b1111))
print(float(0o1234))
print(float(0xFACE3))
````

A `complex` number cannot be converted to `float`:

````python
try:
    print(float(10 + 5j)) # complex > float
except TypeError as e:
    print('TypeError:', e)
````

`bool` and numeric strings convert to `float` as well:

````python
print(float(True), float(False)) # bool > float
print(float('10')) # str > float
print(float('12.345'))
````

### `complex()`

**Form 1:**

```python
complex(x)
```

`complex(x)` — converting a single number to `complex`:

````python
print(complex(10)) # int > complex
print(complex(0b1110))
print(complex(10.5)) # float > complex
print(complex(True), complex(False)) # bool > complex
````

A numeric string also converts to `complex`:

````python
print(complex('10')) # str > complex
print(complex('10.55'))
````

A string written in binary form is not a valid `complex` literal, so it raises a `ValueError`:

````python
try:
    print(complex('0b1010'))
except ValueError as e:
    print('ValueError for 0b1010:', e)
````

**Form 2:**

```python
complex(x, y)
```

`complex(x, y)` — building a complex number from a real and imaginary part:

````python
print(complex(10, 5)) # int > complex
print(complex(0b111, 5))
print(complex(10.5, 5.2)) # float > complex
````

Strings aren't accepted as arguments in this two-value form, so this raises a `TypeError`:

````python
pair1 = ('10', '5')
try:
    print(complex(pair1[0], pair1[1]))
except TypeError as e:
    print('TypeError for', pair1, ':', e)

pair2 = (10, '5')
try:
    print(complex(pair2[0], pair2[1]))
except TypeError as e:
    print('TypeError for', pair2, ':', e)
````

### `bool()`

When converting to the Boolean (`bool`) type, all numbers except `0` are considered `True`, and all strings except empty strings (`''`) are considered `True`.

````python
print(repr(-100), '->', bool(-100))
print(repr(10.999), '->', bool(10.999))
print(repr(0.0001), '->', bool(0.0001))
print(repr(10+5j), '->', bool(10+5j))
print(repr(0b1111), '->', bool(0b1111))
print(repr(0), '->', bool(0))
print(repr('Hello World'), '->', bool('Hello World'))
print(repr(' '), '->', bool(' '))
print(repr(''), '->', bool(''))
````

### `str()`

The `str()` function converts non-string data types into strings:

````python
print(str(10))
print(str(0b11001))
print(str(10.99))
print(str(1 + 0j))
print(str(True), str(False))
````
