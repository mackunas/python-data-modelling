# String Methods

Python provides a set of built-in **string methods** that make it easy to manipulate text.

> **Important:** String methods **do not modify the original string** because strings are **immutable**. Instead, they return a **new string** with the requested changes. If you want to keep the result, assign it to a new variable (or overwrite the original one).

### `str.capitalize()`

The `capitalize()` method returns a new string where:
- the **first character** is converted to uppercase,
- **all remaining characters** are converted to lowercase.

### Example

````python
string = "the quick brown fox jumps over a lazy dog"

print(string.capitalize())
````

### `str.title()`

The `title()` method returns a new string where the **first letter of every word** is converted to uppercase.

### Example

````python
string = "the quick brown fox jumps over a lazy dog"

print(string.title())
````

---

### `str.upper()`

The `upper()` method returns a new string with **all alphabetic characters converted to uppercase**.

### Example

````python
string = "the quick brown fox jumps over a lazy dog"

print(string.upper())
````

---

### `str.lower()`

The `lower()` method returns a new string with **all alphabetic characters converted to lowercase**.

### Example

````python
string = "The Quick Brown Fox Jumps Over A Lazy Dog"

print(string.lower())
````

### `str.swapcase()`

The `swapcase()` method returns a new string with the **case of every letter reversed**:
- uppercase letters become lowercase,
- lowercase letters become uppercase.

### Example

````python
string = "The Quick Brown Fox Jumps Over A Lazy Dog"

print(string.swapcase())
````

---

### `str.isupper()`

The `isupper()` method returns `True` if **all alphabetic characters** in the string are uppercase and there is at least one alphabetic character. Otherwise, it returns `False`.

### `str.islower()`

The `islower()` method returns `True` if **all alphabetic characters** in the string are lowercase and there is at least one alphabetic character. Otherwise, it returns `False`.

### Example

````python
print("Python".isupper())
print("Python".islower())
print("PYTHON".isupper())
print("python".islower())
````

### `str.center(n[, fillchar])`

Centers a string. If the string is shorter than `n`, it is padded with the `fillchar` character.

````python
string = 'The Quick Brown Fox Jumps Over a Lazy Dog'

print('|{0}|'.format(string.center(47)))
print('|{0}|'.format(string.center(47, '*')))
````

### `str.ljust(n[, fillchar])`

Aligns the string to the left. If the string is shorter than `n`, it is padded with the `fillchar` character.

### `str.rjust(n[, fillchar])`

Aligns the string to the right. If the string is shorter than `n`, it is padded with the `fillchar` character.

````python
string = 'The Quick Brown Fox Jumps Over a Lazy Dog'

print('|{0}|'.format(string.ljust(47)))
print('|{0}|'.format(string.rjust(47, '-')))
````

### `str.split(sep=None, maxsplit=-1)`

Returns a list of words from the string using the `sep` separator (by default, a space).

### `str.rsplit(sep=None, maxsplit=-1)`

The returned list can contain at most `maxsplit + 1` elements.

````python
string = '1,2,3,4,5'

print(string.split(','))
print(string.split(',', maxsplit=2))
print(string.rsplit(',', maxsplit=2))
````

````python
string = ' 1,2, ,3, ,4 '
print(string.split(','))

string = '1 2 3 4 5'
print(string.split())

string = ' 1   2   3   4   5 '
print(string.split())
````

### `sep.join(sequence)`

Returns a string created by joining the elements of `sequence`, inserting `sep` between them.

````python
L = ['The', 'Quick', 'Brown', 'Fox', 'Jumps', 'Over', 'a', 'Lazy', 'Dog']

string = ' '.join(L)
print(string)

string = '--'.join(L)
print(string)
````

````python
s = 'The Quick Brown Fox Jumps Over a Lazy Dog'

output = ''.join(r for r in s if r.isupper())
print(output)
````

### `str.replace(old, new[, count])`

The `replace()` method returns a copy of the string where the first `count` occurrences of the specified substring are replaced with another specified substring.

````python
string = 'The quick brown fox jumps over a lazy lazy dog'

print(string.replace('lazy', 'noisy'))
print(string.replace('lazy', 'noisy', 1))
````

### `str.isalnum()`

Returns `True` if all characters are letters or digits.

### `str.isalpha()`

Returns `True` if all characters are letters.

### `str.isdigit()`

Returns `True` if all characters are digits.

````python
string_1 = 'The Quick Brown Fox Jumps Over a Lazy Dog'
string_2 = '1234567890'

print(string_2.isalnum())
print(string_2.isdigit())
print(string_1.isalnum())
print(string_1.isalpha())
````

````python
a = 'Some        string   with    meny          space or tabs'

b = a.split()
b = ''.join(b)

print(b)
print(b.isalnum())

print(''.join(a.split()).isalnum())  # shorter version

print((a.replace(' ', '')).isalnum())
````

### `str.isspace()`

Returns `True` if all characters are whitespace.

````python
s = '    '

print(s.isspace())
````

### `str.strip(characters)`

Returns a copy of the string with both leading and trailing characters removed according to the specified `characters` argument.

### `str.lstrip(characters)`

Returns a copy of the string with characters removed only from the **left side** (beginning) of the string.

### `str.rstrip(characters)`

Returns a copy of the string with characters removed only from the **right side** (end) of the string.

````python
s = '   Hello   '
print(s.strip())

s = ' ***Hello*** '
print(s.strip(' *'))
print(s.strip('*'))
````

````python
s = 'abc***Hello***abc'
print(s.lstrip('a*bc'))
print(s.rstrip('*cba'))
print(s.strip('abc*'))
````

### `str.count(sub[, start[, stop]])`

Returns the number of occurrences of the substring `sub` in the string or in the specified slice (`start` – `stop - 1`).

````python
s = 'ABCABCABCA'

print(s.count('ABC'))      # 3
print(s.count('CA'))       # 3
print(s.count('ABC', 2))   # 2
print(s.count('ABC', 2, 5))# 0
````

### `str.find(sub[, start[, stop]])`

Returns the lowest index where the substring `sub` starts within the range `str[start:stop]`.

If the substring `sub` is not found, returns `-1`.

The method is used when the starting index of a substring needs to be found.

### `str.rfind(sub[, start[, stop]])`

Returns the highest index where the substring `sub` starts within the range `str[start:stop]`.

````python
s = 'First, Second, Third, Third, Fourth'

print(s.find('F'))
print(s.rfind('F'))

print(s.find('Third'))
print(s.rfind('Third'))

print(s.find('Second', 7, 10))
print(s.find('Second', 7, 13))
````

### `str.index(sub[, start[, stop]])`

Returns the lowest index where the substring `sub` starts within the range `str[start:stop]`.

If the substring `sub` is not found, it raises a `ValueError` exception.

### `str.rindex(sub[, start[, stop]])`

Returns the highest index where the substring `sub` starts within the range `str[start:stop]`.

````python
s = 'First, Second, Third, Third, Fourth'

print(s.index('F'))
print(s.rindex('F'))

print(s.index('Third'))
print(s.rindex('Third'))

print(s.index('Second', 7, 10))
print(s.index('Second', 7, 13))
````

### Exception handling example

````python
import datetime as dt

email = input('Enter email address: ')

# Exception handling
try:
    i = email.index('@')
    print('\n"@" index -', i)
except ValueError:
    print('\nWrong email address!')

print(dt.date.today())
````

````python
# Program that outputs the starting indexes of all specified substring occurrences
s = 'ABCABCABCA'

f = input('Enter a substring: ')

i = s.find(f)

if i == -1:
    print('This substring does not exist in the string')

while i != -1:
    print(i)
    i = s.find(f, i + len(f), len(s))
````

# String Comparison

String comparison is performed using comparison operators:

`==`, `!=`, `<`, `>`, `<=`, `>=`

````python
str1 = 'The quick brown fox jumps over a lazy dog'

str2 = input()

print(id(str1))
print(id(str2))

print(str1 == str2)
print(str1 != str2)
print(str1 is str2)
````

````python
str1 = 'The quick brown fox jumps over a lazy dog'

str2 = 'The quick brown fox jumps over a lazy dog'

print(id(str1))
print(id(str2))

print(str1 == str2)
print(str1 is str2)
````

````python
str1 = 'The quick brown fox jumps over a lazy dog'
str2 = 'The Quick Brown Fox Jumps Over a Lazy Dog'

print(str1.lower() == str2.lower())
````

````python
str1 = 'abstract'
str2 = 'abstraction'

print(str1 < str2)
````

### `ord()` Function

The `ord()` function returns the character's position number in the encoding table.

````python
str1 = 'abstract'
str2 = 'absurd'

print(ord('t'))
print(ord('u'))

print(str1 > str2)
````

### `chr()` Function

The `chr()` function returns a character based on its number in the encoding table.

````python
print(chr(65))
print(chr(97))
````

### Membership operators `in` and `not in`

The `in` and `not in` operators are used to check whether a substring exists in a string.

````python
string = 'The quick brown fox jumps over a lazy dog'

print('quick' in string)
print('lazy' not in string)
print('noisy' not in string)
````

## Negative Indexes

```
 0  1  2  3  4  5
 P  y  t  h  o  n
-6 -5 -4 -3 -2 -1
```

`neg_index = n - len(s)`, where `n` is a positive index.

````python
s = 'Python'

for i in range(len(s)):
    print('{}: positive index: {}, negative index: {}'.format(
        s[i], i, i - len(s)
    ))
````

### `sorted()` Function

The `sorted()` function returns a sorted list from the specified iterable object.

````python
s = 'NCMB2RI943JUEA8Y9057'

print(sorted(s))
print(''.join(sorted(s)))
````

### `reversed()` Function

The `reversed()` function returns a reverse iterator of the specified sequence.

````python
s = 'Python'

rev = reversed(s)

print(rev)
````

````python
s = 'Python'

rev = reversed(s)

for i in rev:
    print(i)
````

````python
s = 'Python'

rev = ''.join(reversed(s))

print(rev)
````
