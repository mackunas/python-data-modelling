## Programming style (PEP 8)
Most programming languages can be written (or, more accurately, formatted) in different styles. Some styles are easier to read than others. Writing code that is easy for others to understand is a good practice, and using a consistent coding style helps make programs more readable.

Most Python projects follow the style guidelines defined in **PEP 8**, which promote clear and readable code. Some of the most important recommendations are:

- Use **4 spaces** for indentation, and do not use tabs.
- Use blank lines to separate functions, classes, and larger blocks of code.
- Whenever possible, write comments on a separate line.
- Use **docstrings** to document functions, classes, and modules.
- Use spaces around operators and after commas, but not inside parentheses:

```python
a = f(1, 2) + g(3, 4)
```

- Use consistent naming conventions:
  - **Classes:** `CamelCase`
  - **Functions and methods:** `lower_case_with_underscores`
  - Always use `self` as the first parameter of instance methods.

## Data Type conversion
A data type can be converted using the functions `int()`, `float()`, `bool()`, `complex()`, and `str()`.

### `int()`

Converting a `float` to `int` truncates the decimal part:

````python
print(int(10.999))  # float > int
````

A `complex` number cannot be converted to `int`:

````python
try:
    print(int(10 + 5j)) # complex > int
except TypeError as e:
    print('TypeError:', e)
````

`bool` converts to `0` or `1`, and a numeric string converts directly:

````python
print(int(True), int(False)) # bool > int
print(int('20')) # str > int
````

Strings that aren't plain base-10 integers raise a `ValueError`:

````python
for bad in ['10.5', '0b1111']:
    try:
        print(int(bad))
    except ValueError as e:
        print('ValueError for', bad + ':', e)
````

### `float()`

`int` values, including those written in binary, octal, or hex, all convert to `float`:

````python
print(float(100)) #int > float
print(float(0b1111))
print(float(0o1234))
print(float(0xFACE3))
````

A `complex` number cannot be converted to `float`:

````python
try:
    print(float(10 + 5j)) # complex > float
except TypeError as e:
    print('TypeError:', e)
````

`bool` and numeric strings convert to `float` as well:

````python
print(float(True), float(False)) # bool > float
print(float('10')) # str > float
print(float('12.345'))
````

### `complex()`

**Form 1:**

```python
complex(x)
```

`complex(x)` — converting a single number to `complex`:

````python
print(complex(10)) # int > complex
print(complex(0b1110))
print(complex(10.5)) # float > complex
print(complex(True), complex(False)) # bool > complex
````

A numeric string also converts to `complex`:

````python
print(complex('10')) # str > complex
print(complex('10.55'))
````

A string written in binary form is not a valid `complex` literal, so it raises a `ValueError`:

````python
for bad in ['0b1010']:
    try:
        print(complex(bad))
    except ValueError as e:
        print('ValueError for', bad + ':', e)
````

**Form 2:**

```python
complex(x, y)
```

`complex(x, y)` — building a complex number from a real and imaginary part:

````python
print(complex(10, 5)) # int > complex
print(complex(0b111, 5))
print(complex(10.5, 5.2)) # float > complex
````

Strings aren't accepted as arguments in this two-value form, so this raises a `TypeError`:

````python
for bad in [('10', '5'), (10, '5')]: # str > complex
    try:
        print(complex(*bad))
    except TypeError as e:
        print('TypeError for', bad, ':', e)
````

## Mutable and Immutable Objects

A **mutable object** is an object that can change its value while keeping the same `id()`.

An **immutable object** is an object with a fixed value.

If we want to change the value of an immutable object, a new object must be created.

### The `id()` Function

Every Python object is stored in a specific memory location.

The `id()` function returns the memory location of a given object. The address is returned as a base-10 number.

This function can be used to check whether objects refer to the same memory location, meaning they refer to the same data in memory.

### The `is` Keyword

`is` is a keyword that compares two objects and returns `True` if both objects have the same memory address.

`is not` returns `False` if both objects have the same memory address.

### Immutable Objects

- Numbers (`int`, `float`, `decimal`, `boolean`, `complex`)
- Strings
- Tuples
- Frozen sets

Reassigning a variable to a new value creates a new object, so `id(a)` changes:

````python
a = 10
print('a =', a, ' id(a)=', id(a))
a = 20
print('a =', a, ' id(a)=', id(a))
````

When several variables are assigned from the same object, they all share the same `id` and `is` returns `True`:

````python
a = 1000
b = a
c = a
print('id(a)=', id(a))
print('id(b)=', id(b))
print('id(c)=', id(c))
````

````python
print('a is b ->', a is b)
print('a is c ->', a is c)
print('b is c ->', b is c)
````

Small integers are cached by Python, so two variables holding the same small value may be `is`-identical. But once a variable is reassigned to a different value, `is` no longer holds:

````python
a = 10
b = a
print('a is b ->', a is b)
a = 1000
print('a is b ->', a is b)
````

Mutable objects include:

- Lists
- Sets
- Dictionaries

A **list** is a collection of values of any data type, separated by commas and enclosed in square brackets.

Lists are mutable: changing an element in place does not change the list's `id`:

````python
my_list = [10, 20, 30, 40]
print(id(my_list))
my_list[2] = 3000
print(id(my_list))
print(my_list)
````

Assigning a list to another variable does not copy it — both names refer to the same object, so modifying one affects the other:

````python
list1 = [10, 20, 30, 40]
list2 = list1
print('list1 is list2 ->', list1 is list2)
list1[0] = 100.52
print('list1 is list2 ->', list1 is list2)
print('list1 =', list1)
print('list2 =', list2)
````

Two different lists that happen to hold equal values are still two distinct objects:

````python
list1 = [10, 20, 30, 40]
list2 = [10, 20, 30, 40]
print('id(list1)=', id(list1))
print('id(list2)=', id(list2))
````

Two separate empty lists are equal in value (`==`) but are not the same object (`is`):

````python
x = []
y = []
print('x == y ->', x == y)
print('x is y ->', x is y)
print('x is not y ->', x is not y)
````

Tuples are immutable: trying to change an element raises a `TypeError`:

````python
t = (1, 2, 3) # tuple
try:
    t[0] = 99
except TypeError as e:
    print('TypeError:', e)
````

Sets and dictionaries are mutable too, so items can be added after creation. A `frozenset` is the immutable counterpart of a `set`:

````python
s = {1, 2, 2, 3} # set
s.add(4)
print('set:', s)
d = {'a': 1, 'b': 2} # dictionary
d['c'] = 3
print('dict:', d)
fs = frozenset([1, 2, 2, 3])
print('frozenset:', fs)
````

## Input and Output

`eval()` isn't limited to `True`/`False` — it evaluates any Python expression the user types in. For example, a full arithmetic expression is evaluated, not just read as text:

## Formatted output
### The String Formatting Method `.format()`

The `format()` method inserts values into a string using `{}` placeholders.

Ways to use it:
- `{}` uses values in order.
- `{0}`, `{1}`, etc. use positions.
- `{name}` uses named arguments.

````python
first_name = 'John'
last_name = 'Smith'
salary = 2500
print('Hello {} {}. Your salary is {}'.format(first_name, last_name, salary))
print('Hello {0} {1}. Your salary is {2}'.format(first_name, last_name, salary))
print('Hello {1} {0}. Your salary is {2}'.format(first_name, last_name, salary))
print('Hello {a} {b}. Your salary is {c}'.format(a=first_name, b=last_name, c=salary))
print('Hello {} {}. Your salary is {}'.format('John', 'Smith', 2500))
````

### Output Formats

- `:d` – integer (`int`)
- `:f` – floating-point number (`float`)
- `:e` – exponential notation
- `:s` – string (`str`)

````python
n = 458.541315
try:
    print('{:d}'.format(n))  # wrong: n is float
except ValueError as e:
    print('ValueError:', e)
print('{:d}'.format(int(n)))
print('{:f}'.format(n))
print('{0:15f}'.format(n))
print('{0:15e}'.format(n))
print('{0:15.3f}'.format(n))
````

### Alignment

- `<` – align to the left
- `>` – align to the right
- `^` – align to the center

````python
print('{0:15f}{1}'.format(n, '|'))
print('{0:<15f}{1}'.format(n, '|'))
print('{0:>15f}{1}'.format(n, '|'))
print('{0:^15.3f}{1}'.format(n, '|'))
print('{:s}'.format('hello'))
````
