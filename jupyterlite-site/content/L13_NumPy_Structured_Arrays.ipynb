# Lecture 13 — Structured Arrays

A structured array is the standard NumPy way of storing heterogeneous tabular data.

## Data types

| Numpy data type symbols/abbreviations | Description | Example |
|---|---|---|
| 'i' ('i1','i2','i4','i8') | Signed integer | `np.dtype('i4')==np.int32` |
| 'u' ('u1','u2','u4','u8') | Unsigned integer | `np.dtype('u1')==np.uint8` |
| 'f' ('f2','f4','f8') | Floating point | `np.dtype('f8')==np.int64` |
| 'c' ('c8','c16') | Complex floating point | `np.type('c16')==np.complex128` |
| 'S','a' | zero-terminated bytes (not recommended) | `np.dtype('S5')` |
| 'U' | Unicode encoded string | `np.dtype('U')=np.str_` |

## Ways to create and index structured arrays

### List of tuples: one tuple - one field

In this method, the structured data type is defined as a list of tuples. One tuple corresponds to only one data field.

Tuples have the form (field name, data type, subarray shape), where the subarray shape parameter is optional.

The field name parameter can be a string, or, if field titles are used, a tuple. The optional subarray shape parameter is a shape of integers defining the arrays that will be stored in this field.

Let's create a structured array based on the table:

| Row index | name | age | weight |
|---|---|---|---|
| 0 | Jack | 19 | 81.5 |
| 1 | Alice | 20 | 51 |
| 2 | Bob | 20 | 70.2 |
| 3 | Cathy | 21 | 52 |

````python
import numpy as np

s_arr = np.array([('Jack', 19, 81.5),
                   ('Alice', 20, 51),
                   ('Bob', 20, 70.2),
                   ('Cathy', 21, 52)],
                  dtype = [('name', 'U8'), ('age', 'i4'), ('weight', 'f8')])

print('s_arr.dtype')
print(s_arr.dtype, end='\n\n') # the array's data type
````

````python
print('s_arr')
print(s_arr, end='\n\n')
````

````python
print('s_arr rows')
for i in s_arr: # output as a column
    print(i)
````

A structured type can be defined in advance as a variable and used to create the array:

````python
s_arr_type = np.dtype([('name', 'U8'), ('age', 'i4'), ('weight', 'f8')])

s_arr = np.array([('Jack', 19, 81.5),('Alice', 20, 51),('Bob', 20,
70.2),('Cathy', 21, 52)], dtype=s_arr_type)
print(s_arr.dtype)
````

### Indexing

````python
print("s_arr['name']")
print(s_arr['name'])
print("s_arr['name'][1]")
print(s_arr['name'][1])
print("s_arr[1]['name']")
print(s_arr[1]['name'])
````

Rows can be sliced using ranges or negative indices, just like a normal NumPy array:

````python
print("s_arr['name'][1:]")
print(s_arr['name'][1:])
print('s_arr[-1]')
print(s_arr[-1])
print('s_arr[1:3]')
print(s_arr[1:3])
````

Several fields can be selected at once with a list of field names, and boolean masks work the same way as with ordinary arrays:

````python
print("s_arr[['name','age']][2]")
print(s_arr[['name','age']][2])
print("s_arr[s_arr['weight']<70]['name']")
print(s_arr[s_arr['weight']<70]['name'])
# Array s_arr
# ('Jack', 19, 81.5)
# ('Alice', 20, 51.)
# ('Bob', 20, 70.2)
# ('Cathy', 21, 52.)
````

A structured array can be created without specifying column names, in which case names are automatically assigned to the array's columns (f0, f1,...,fn).

Using them as a structure index, elements of the same type can be specified.

````python
arr = np.array([(1, 'First', 0.5, 1+2j), (2, 'Second', 1.3, 2-2j),
(3, 'Third', 0.8, 1+3j)], dtype=('i2, U6, f4, c8'))

print('arr.dtype')
print(arr.dtype, end='\n\n')
````

````python
print('arr rows')
for i in arr:
    print(i)
print()
````

Even with auto-generated field names (`f0`, `f1`, ...), individual fields and multi-field selections work exactly the same as with named fields:

````python
print("arr['f1']")
print(arr['f1'], end='\n\n')
````

````python
print("arr[['f0','f3']][2]")
print(arr[['f0','f3']][2])
````

Column names can be created separately:

````python
arr = np.array([(1, 'First', 0.5, 1+2j), (2, 'Second', 1.3, 2-2j),
(3, 'Third', 0.8, 1+3j)], dtype=('i2, U6, f4, c8'))

arr.dtype.names = ('id', 'order', 'value', 'complex') # assigning names to the columns

print(arr['complex'])
````

Sometimes it is necessary to create a structured array with specified field names and other parameters, filled with identical elements. Such arrays are usually created for further filling in:

````python
arr_type = [('name', 'U10'),('age', 'i4'),('weight', 'f8')]

arr=np.ones(4,dtype=arr_type) # np.ones

print('arr.shape')
print(arr.shape)
print('arr.dtype')
print(arr.dtype)
print('arr')
print(arr)
````

Data can then be written into it:

````python
name = np.array(['Jack','Alice','Bob','Cathy'])
age = np.array([19,20,20,21])
weight = np.array([81.5,51,70.2,52])

arr_type = [('name', 'U10'),('age', 'i4'),('weight', 'f8')]

arr=np.ones(4,dtype=arr_type)

arr['name']=name
arr['age']=age
arr['weight']=weight
print(arr)
````

### Another way of assigning values

````python
# List of tuples of values
L = [('Jack',19,81.5),('Alice', 20, 51.),('Bob',20,70.2), ('Cathy',21,52)]

arr_type = [('name', 'U10'),('age', 'i4'),('weight', 'f8')]

arr=np.ones(4,dtype=arr_type)

arr[:] = L

print(arr)
````

### Dictionary of field parameter arrays

This is the most flexible form of specification, since it allows controlling the byte offsets of fields and the size of the structure's elements.

Two dictionary keys are required — "names" and "formats":

```
dtype = {'names':['member name 1',...,'member name n'],'formats':['data type 1',...,'data type n']}
```

where:
- member name – used as the table column name;
- data type – a Python or Numpy data type.

````python
cars = np.array([('mazda', 4, 3573.5),('tesla', 2, 4226.81),('bmv', 8,
3489.21)],
                 dtype =
{'names':['model','quantity','value'],'formats':['U10','i4','f4']})

for i in cars:
    print(i)
````

````python
print("mean(arr['X'])")
print(np.mean(arr['X']), end='\n\n')
````

````python
print("max(arr['Y'])")
print(np.max(arr['Y']), end='\n\n')
````

## Sorting

### The sort() function

`sort(a, axis=-1, kind='quicksort', order=None)`

Parameters:

- `a` – the array;
- `axis` – an integer (optional). Specifies the axis along which the elements are sorted. If it is equal to `None`, a multidimensional array is flattened into a one-dimensional one. The default value is (-1), which corresponds to sorting along the last axis of the array.
- `kind` – a string (optional). One string from the set {'quicksort', 'mergesort', 'heapsort', 'stable'}, which specifies the sorting algorithm. The 'stable' value automatically selects a stable, best algorithm for the data type of the array being sorted. The default value is 'quicksort'.
- `order` – a string or a list of strings (optional). Allows setting the sort order of a structured array.

````python
a = np.random.randint(0, 50, 10)
print('array a')
print(a)
````

````python
print('sort(a)')
print(np.sort(a))
````

````python
a = np.random.randint(10, 100, size = (4,4))
print('array a')
print(a, end='\n\n')
````

````python
print('sort(a) - sorts along axis=1 (each row)')
print(np.sort(a), end='\n\n')
````

````python
print('sort(a, axis=0) - sorts each column')
print(np.sort(a,axis=0))
````

````python
arr = np.array([('Jack', 19, 81.5),('Alice', 21, 51),('Bob', 20,
70.2),('Cathy', 20, 52)],dtype = [('name', 'U8'), ('age', 'i4'),
('weight', 'f8')])

print('field names')
print(arr.dtype.names, end='\n\n')
````

````python
print('unsorted')
for i in arr:
    print(i)
print()
````

The `order` parameter sorts by one field, or a list of fields as tie-breakers when the first field has equal values:

````python
print("sorted by 'name'")
for i in np.sort(arr,order='name'):
    print(i)
print()
````

````python
print("sorted by 'age'")
for i in np.sort(arr,order='age'):
    print(i)
print()
````

````python
# sorts by 'age', then by 'weight', if the 'age' values are equal
print("sorted by 'age', then 'weight'")
for i in np.sort(arr,order=['age','weight']):
    print(i)
````

This function has an equivalent `ndarray` class method `a.sort()`. However, unlike the function, this method does not return a copy, but changes the original array:

````python
a = np.array([[5.,0.,1.,3.,2.],
              [3.,2.,9.,4.,8.],
              [1.,6.,3.,1.,3.]])
print('array a before')
print(a, end='\n\n')

(a[:,0]).sort() # changes the first column of the array
print('array a after sorting column 0 in place')
print(a)
````

### The argsort() function

`argsort(a, axis=-1, kind='quicksort', order=None)`

The function returns the indices that sort the array.

````python
score = np.array([75, 62, 50, 12, 90, 48, 83])
name = np.array(['Eva', 'Ben', 'Charlie', 'Danny', 'Eden', 'Fanny', 'George'])

print('score')
print(score)
````

````python
print('sort(score)')
print(np.sort(score))
````

````python
print('argsort(score)') # indices that sort the array score
print(np.argsort(score))
````

````python
print('score[argsort(score)]') # the array is sorted using the index array
print(score[np.argsort(score)])
````

The same index array returned by `argsort()` can be used to reorder a *different* array in matching order — here, pairing each score with its student's name, then taking the top 3:

````python
# print the 3 top-performing students
# ['Eva' 'Ben' 'Charlie' 'Danny' 'Eden' 'Fanny' 'George']
# [ 75 62 50 12 90 48 83 ]

print('name[argsort(score)]')
print(name[np.argsort(score)])
````

````python
print('top 3 students by score')
print((name[np.argsort(score)][::-1])[:3])
````

````python
# Sorts the columns so that the elements of the first row are arranged in increasing order.
a = np.array([[ 5.,0.,1.,3.,2.],
              [-3.,2.,9.,4.,8.],
              [-1.,6.,3.,1.,3.]])
print('array a')
print(a, end='\n\n')
````

````python
print('a[0]')
print(a[0], end='\n\n')
print('argsort(a[0])')
print(np.argsort(a[0]), end='\n\n')
````

````python
b = a[:, np.argsort(a[0])]
print('b = a reordered by argsort(a[0]) columns')
print(b, end='\n\n')
````

`argsort()` also accepts an `axis` parameter, just like `sort()`. Sorting along `axis=0` treats each column independently — different from sorting the transposed array's first row:

````python
print('argsort(a[0])')
print(np.argsort(a[0]), end='\n\n')
````

````python
print('argsort(a[0], axis=0)')
print(np.argsort(a[0],axis=0), end='\n\n')
````

````python
print('argsort(a.T[0])') # or np.argsort(a[:,0])
print(np.argsort(a.T[0]))
````

````python
arr = np.array([(0,12),(0,11),(1,55),(0,21),(0,87),(1,54),(1,21)],
dtype = np.dtype([('a', int), ('b', int)]))
print('unsorted (index, row)')
for i,j in zip(range(len(arr)),arr):
    print(i,j)
print()
````

`argsort(order=[...])` on a structured array sorts lexicographically by the given field order — first by `'a'`, then by `'b'` as a tie-breaker (or the reverse, if the field order is reversed):

````python
#sorting the structured array in lexicographic order
ind = np.argsort(arr, order = ['a', 'b'])
print("sorted by 'a' then 'b' (index, row)")
for i,j in zip(ind,arr[ind]):
    print(i,j)
print()
````

````python
print("sorted by 'b' then 'a'")
for i in arr[np.argsort(arr, order = ['b', 'a'])]:
    print(i)
````

### The lexsort() function

`lexsort(keys, axis=-1)`

The function performs sorting according to the specified keys (lexicographic sorting). Sorting is first performed by the key that is specified last.

````python
first_name = np.array(['Bob', 'Jane', 'Steve', 'Bill', 'Barbara'])
last_name = np.array(['Jones', 'Arnold', 'Arnold', 'Jones', 'Walters'])

ind1 = np.lexsort((first_name,last_name)) # sorts by last_name, then by first_name

print('sorted by last_name, then first_name')
for i,j in zip(last_name[ind1],first_name[ind1]):
    print(i,j)
````

Swapping the order of the keys passed to `lexsort()` changes which field is the primary sort key — the *last* key given is the primary key:

````python
ind2 = np.lexsort((last_name,first_name)) # sorts by first_name, then by last_name

print('sorted by first_name, then last_name')
for i,j in zip(first_name[ind2],last_name[ind2]):
    print(i,j)
````

### The partition() function

`partition(a, kth, axis=-1, kind='introselect', order=None)`

The function returns a copy of the original array whose elements are divided according to the specified value. The `kth` parameter specifies the index of an element in the sorted copy of the original array; however, in the returned array only this element retains its position — all elements of smaller value are moved to the left, and elements of greater or equal value to the right, but the order of the left-hand and right-hand elements is not defined in any way.

````python
a = np.random.randint(0,15, 15)
print('array a')
print(a)
````

````python
print('partition(a, 4)')
print(np.partition(a,4))
````

````python
print('sort(a) for comparison')
print(np.sort(a))
````

````python
a = np.array([[9,10,11,12,4],[15,6,17,8,6],[11,4,35,6,5],[1,23,2,4,57]])
print('array a')
print(a, end='\n\n')
````

````python
print('partition(a, 2) - default axis=-1 (each row)')
print(np.partition(a,2), end='\n\n')
````

`axis=0` partitions each column instead of each row; `axis=None` flattens the array first, so the k-th value is the k-th smallest across all elements:

````python
print('partition(a, 1, axis=0) - each column')
print(np.partition(a,1,axis=0), end='\n\n')
````

````python
print('partition(a, 5, axis=None) - flattened')
print(np.partition(a,5,axis=None))
````

This function has an equivalent `ndarray` class method `a.partition()`. However, unlike the function, this method does not return a copy, but changes the original array:

````python
a = np.array([[9,10,11,12,4],[15,6,17,8,6],[11,4,35,6,5],[1,23,2,4,57]])
print('array a before')
print(a)
print()
a[1,:].partition(2) # changes the second row
print('array a after partitioning row 1 in place')
print(a)
````

### The argpartition() function

`argpartition(a, kth, axis=-1, kind='introselect', order=None)`

The function returns an array of the indices of the original array's elements, partitioned according to the specified value.

````python
arr = np.array([(0,12),(0,11),(1,5),(0,2),(0,87),(1,154),(1,21),(0,19)],
dtype = np.dtype([('a', int), ('b', int)]))

print('unsorted rows')
for i in arr:
    print(i)
print()
````

````python
print("argpartition(arr, 2, order='b')")
print(np.argpartition(arr,2,order='b'))
````

## Searching

### The argmax() function

`argmax(a, axis=None, out=None)`

The function returns the index of the largest value along the specified axis. When several elements correspond to the maximum value, the index of the first element is returned.

### The unravel_index() function

`unravel_index(indices, shape, order='C')`

The function converts a flat array of indices into a tuple of index arrays.

Parameters:
- `indices` – indices of elements when the array is flattened into a one-dimensional array.
- `shape` – the shape of the multidimensional array.

````python
v = np.array([3,2,3,4,5,2,5,5,2])
print(np.argmax(v))
````

````python
a = np.array([[ 5.,0.,1.,3.,2.],
              [-3.,2.,9.,4.,3.],
              [-1.,6.,3.,1.,8.]])
print('array a')
print(a, end='\n\n')
````

````python
print('argmax(a, axis=0) - index of max in each column')
print(np.argmax(a,axis=0))
````

````python
print('argmax(a, axis=1) - index of max in each row')
print(np.argmax(a,axis=1))
````

````python
print('argmax(a) - index into the flattened array')
print(np.argmax(a), end='\n\n')
````

`unravel_index()` converts that flat index back into a (row, column) coordinate for the original shape:

````python
ind=np.unravel_index(np.argmax(a),a.shape)
print('unravel_index(argmax(a), a.shape)')
print(ind)
print('a[ind]')
print(a[ind])
````

````python
arr = np.array([(0,2),(0,87),(1,154),(1,21),(0,19)],
dtype = np.dtype([('a', int), ('b', int)]))

print('rows')
for i in arr:
    print(i)
print()
````

````python
print("argmax(arr['b'])")
print(np.argmax(arr['b']))
````

````python
print("arr[argmax(arr['b'])]")
print(arr[np.argmax(arr['b'])])
````

### The argwhere() function

`argwhere(a)`

The function returns the indices of the array's non-zero elements.

````python
v = np.array([ 0.,1.,0.,2.,3.,0.,5.,0.])

print('array v')
print(v, end='\n\n')
````

````python
a = np.array([[5.,0.,0.],
              [0.,2.,0.],
              [1.,6.,3.]])

print('array a')
print(a, end='\n\n')
````

````python
print('nonzero(v)')
print(np.nonzero(v))
````

### The where() function (condition only)

If only a condition is specified in the function, the function returns only sets of the array elements' indices (like `nonzero()`).

````python
v = np.array([ 0.,1.,0.,2.,3.,0.,5.,0.])
print('array v')
print(v)
````

````python
print('where(v)')
print(np.where(v), end='\n\n')
````

````python
print('v[where(v)]')
print(v[np.where(v)], end='\n\n')
````

`where()` with a single argument behaves just like `nonzero()` — it returns the indices, which is why it also works the same way on 2-D arrays and with comparisons:

````python
a = np.array([[5.,0.,0.],
              [0.,2.,0.],
              [1.,6.,3.]])
print('array a')
print(a, end='\n\n')
````

````python
print('where(a)')
print(np.where(a))
````

````python
print('where(a>2)')
print(np.where(a>2))
````

````python
print('a[where(a>2)]')
print(a[np.where(a>2)])
````

### The searchsorted() function

`searchsorted(a, v, side='left', sorter=None)`

The function returns the indices at which the specified elements should be inserted in order to preserve the sort order.

Parameters:
- `a` – a one-dimensional array. This array may be unsorted, but the indices of the sorted array are shown.
- `v` – a one-dimensional array. Elements that need to be inserted into array `a`.
- `side` – a string from the set {'left', 'right'} (optional). 'left' – the function returns the index counting from the left, 'right' – returns the index counting from the right.
- `sorter` – an optional array of integer indices that sorts array `a` in increasing order (e.g. the result of `argsort()`).

````python
a = np.array([2,5,4,1,6,8,7,9,2,5,4,1])
print('array a')
print(a)
````

````python
print('searchsorted(a, 3)')
print(np.searchsorted(a,3))
````

````python
print('sort(a) for reference')
print(np.sort(a))
````

### The counting function count_nonzero()

`count_nonzero(a, axis=None)`

The function returns the number of non-zero elements in the array.

````python
a = np.array([[0, 1, 7, 0],
              [3, 0, 2, 19]])

print('count_nonzero(a, axis=0)')
print(np.count_nonzero(a,axis=0))
````

````python
print('count_nonzero(a, axis=1)')
print(np.count_nonzero(a,axis=1))
````

````python
print('count_nonzero(a)')
print(np.count_nonzero(a))
````

````python
print('count_nonzero(a>3)')
print(np.count_nonzero(a>3), end='\n\n')
````

`count_nonzero()` also works on string arrays — an empty string `''` counts as "falsy", just like the number 0:

````python
a = np.array(['','A','a','','B','b','C','',''])
print('array a (strings)')
print(a)
````

````python
print('count_nonzero(a)')
print(np.count_nonzero(a))
````

````python
print("count_nonzero(a > 'B')")
print(np.count_nonzero(a >'B'))
````
