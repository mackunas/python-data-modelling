# Lecture 6 — Dictionaries

A dictionary is a sequence of data that has a key:

```
{key1: value, key2: value, ... }
```

Dictionaries are indexed by keys.

A key can be any immutable data type. Strings and numbers can be used as keys.
Tuples can be used as keys if they consist only of strings, numbers, or tuples.
If a tuple consists of mutable objects (e.g. lists), it cannot be used as a key.
Keys must be unique.

A dictionary can be created using the `{}` operator or the `dict()` function:

## Creating dictionaries

There are several ways to create a dictionary. The simplest is an **empty dictionary**, created either with `{}` or with the `dict()` constructor:

````python
# Empty dictionary
d1 = {}
d2 = dict()

print(type(d1), d1)
print(type(d2), d2)
````

A dictionary can also be created directly as a literal, listing `key: value` pairs separated by commas:

````python
# key: value
d1 = {1:'a', 2:'b', 3:'c'}
print(d1)
````

If the same key is written twice in a literal, the **later value overwrites the earlier one** — the key `100` ends up mapped to `'ABCD'`, not `'A'`:

````python
d2 = {100:'A', 200:'B', 300:'C', 100:'ABCD'} # two identical keys
print(d2)
````

A key does not have to be a single value — it can also be a **tuple**, as long as the tuple itself only contains immutable elements:

````python
d3 = {('a','b'):100, ('5','5'):200} # keys of tuple type
print(d3)
````

When all the keys are valid Python identifiers (i.e. they could be variable names), a dictionary can be created by passing them as **keyword arguments** to `dict()`. Note that keys created this way are always strings:

````python
# Key type - str
pop = dict(Vilnius=540919, Kaunas=286763, Klaipėda=147898)
empl = dict(manager='Smith', salesman='Johnson')
print(pop)
print(empl)
````

### Creating a dictionary from another sequence

`dict()` can also build a dictionary out of a **sequence of two-element items** — each item becomes one `key: value` pair. The outer sequence can be a list, a tuple, or a set, and each inner pair can itself be a list, a tuple, or a set of exactly two elements.

**From a list of pairs and a tuple of pairs:**

````python
# A dictionary made from a list
d1 = dict( [['a','b'], ['c','d']] ) # list of lists
d2 = dict( [('A','B'), ('C','D')] ) # list of tuples
d3 = dict( [{'e','f'}, {'E','F'}] ) # list of sets (!!! ASKED AI is it safe and add comment if unsafe and ADDED comment below)
d4 = dict( (['a','b'], ['c','d']) ) # tuple of lists
d5 = dict( (('A','B'), ('C','D')) ) # tuple of tuples
d6 = dict( ({'e','f'}, {'E','F'}) ) # tuple of sets
print('{}\n{}\n{}\n{}\n{}\n{}'.format(d1, d2, d3, d4, d5, d6))
````

**Important:** Although sets can technically be used as pairs when each set contains exactly two elements, this is unsafe because sets are unordered. Python does not guarantee which element becomes the dictionary key and which becomes its value. Therefore, use lists or tuples when creating dictionaries from pairs.

A **string of exactly two characters** also counts as a two-element sequence, so a list of such strings works too — the first character becomes the key, the second the value:

````python
d4 = dict( ['aA', 'bB', 'cC'] )
print(d4)
````

**From a set of pairs.** Since a set is unordered, be aware that the order of items — and therefore which value ends up assigned to which key when duplicates could arise — is not guaranteed:

````python
# A dictionary made from a set
d1 = dict( {('A','B'), ('C','D')} ) # set of tuples
d2 = dict( {'aA', 'bB', 'cC'} )
print('{}\n{}'.format(d1,d2))
````

A dictionary can be created with the help of the `dict.fromkeys(sequence[, value])` method.

The method returns a dictionary with the specified keys and the specified value.

`sequence` - a required parameter, an iterable, specifying the keys of the new dictionary.

`value` - an optional parameter, the value for all keys. If the value is not specified, the default value is None.

Below, `d1` shows that duplicate keys in the input sequence just collapse into one entry, `d2` and `d3` assign the same fixed value to every key, and `d4` shows what happens when no value is given at all:

````python
d1 = dict.fromkeys([1,1,2,2,3],100)
print(d1)
````

````python
d2 = dict.fromkeys(['a','b'],0.)
print(d2)

d3 = dict.fromkeys((100,200,300),'A')
print(d3)
````

If `value` is omitted, every key is mapped to `None`:

````python
d4 = dict.fromkeys(range(5))
print(d4)
````

Dictionary elements are indexed by keys.

````python
d = {'one':100, 'two':200, 'three':300}
print(d['one'])
````

Trying to access a key that doesn't exist raises a `KeyError` — this is different from a list, where an out-of-range index raises `IndexError`:

````python
print(d['four'])
````

A dictionary is a mutable object.

A new dictionary pair can be created by specifying the key and the value assigned to it:

````python
d = {'one':100, 'two':200, 'three':300}
d['four'] = 400 
d['two'] = 2000000 # the value is changed
print(d)
````

The next example builds up a dictionary called `person` step by step, adding one key at a time — a common pattern when the data becomes available piece by piece:

````python
# The program creates the dictionary "person"
s = 'Vardas Pavardė Vilnius VilniusTech 9 9 8 10 10 7 9 10'.split()
print(s)

person = {}
person['fName'] = s[0]
person['lName'] = s[1]
person['city'] = s[2]
person['university'] = s[3]
person['marks'] = [int(x) for x in s[4:]]

print(person)
````

````python
# Dictionary "person"
# Access the 4th mark, add a new grade to the marks list, and print all marks
print(person['marks'][3],'\n')
person['marks'].append(int(input('Enter a grade: ')))

for i in person['marks']:
 print(i, end=' ')
````

````python
# Dictionary "person"
for k in person:
 print('{0:14s} {1}'.format(k, person[k]))
````

## Membership testing

The functions `in dict` and `not in dict` check whether the specified key exists in the dictionary.

````python
d = {100:'one', 200:'two', 300:'three'}
print(100 in d)
print(500 in d)
print('one' in d)
````

````python
if 400 not in d:
 d[400] = 'four'
 print(d)
````

## Arithmetic operations and comparisons

The arithmetic operations "+" and "*" are not allowed with dictionaries.

Only the `==` operator can be used for comparison.

````python
d1 = {100:'A', 200:'B'}
d2 = {300:'C', 400:'D'}
d3 = {200:'B', 100:'A'}

print(d1 == d2)
print(d1 == d3)
````

### min(), max()

When `min()` and `max()` are applied directly to a dictionary, they compare only the dictionary keys, not the values.

- The `min()` function returns the smallest key.
- The `max()` function returns the largest key.

````python
d = {'x': 100, 'b': 100, 'e': 200, 'a': 300, 'h': 100, 'd': 100}
print(min(d))
print(max(d))
````

## Dictionary methods


`dict.keys()` - returns a view object of the dictionary's keys.

`dict.values()` - returns a view object of the dictionary's values.

````python
empl = {'salesman': 'Johnson', 'salary': 1500, 'age': 30}

print(empl.keys())
print(empl.values())
````

`dict.keys()` (and `values()`/`items()`) return **view objects**, not independent lists — they stay live and automatically reflect later changes to the dictionary:

````python
d = {1: 100, 2: 200, 3: 300, 4: 100}

keys = d.keys()
print(keys)

del d[3]
print(keys)
````

````python
empl = {'salesman': 'Johnson', 'salary': 1500, 'age': 30}

for v in empl.values():
 print(v)
````

````python
d = {1: 100, 2: 200, 3: 300, 4: 100}
v_sum = sum(d.values())

print(v_sum)
````

View objects don't support list methods like `.sort()` directly:

````python
d = {1: 100, 2: 200, 3: 300, 4: 100}

d_val = d.values()
print(d_val)

d_val.sort() # a method of the list type object
````

Dictionary values can be sorted using the `sorted()` function. Another way is to first convert the dictionary values into a list using `list()` and then sort the list using the `sort()` method.

````python
d = {1: 100, 2: 200, 3: 300, 4: 100}

# Method 1: Using sorted()
sorted_list = sorted(d.values())
print(sorted_list)

# Method 2: Using list() and sort()
d_val = list(d.values())
d_val.sort()
print(d_val)
````

````python
d = {'a': 100, 'b': 200, 'c': 300}
print(200 in d.values())
````

`dict.copy()` - creates a copy of a dictionary

Plain assignment (`d2 = d1`) does **not** copy the dictionary — `d2` just becomes another name for the same object in memory, so changing `d2` also changes `d1`:

````python
d1 = {'a':100, 'b':200, 'c':300}
d2 = d1

print(id(d1))
print(id(d2))

d2['d'] = 400

print(d1)
````

`dict.copy()` creates an actual independent copy, with its own memory address, so changes to `d2` no longer affect `d1`:

````python
d1 = {'a':100, 'b':200, 'c':300}
d2 = d1.copy()

print(id(d1))
print(id(d2))

d2['d'] = 400

print(d1)
````

`dict.get(key[, x])` - returns the value indicated by the key `key`. If there is no such key, returns `x` or `None`.

Note the last line below is a deliberate bug: `coord[w]` uses the bare (undefined) name `w` instead of the string `'w'`, so it raises a `NameError` — it is **not** the same as `coord['w']`, which would raise a `KeyError` since `'w'` isn't a key in `coord`:

````python
coord = {'x': 0.0, 'y': 0.0, 'z': 10.}

print(coord.get('z'))
print(coord.get('w'))
print(coord.get('w', 'There is no such key in the dictionary'))

print(coord[w])
````

`get()` is very handy for counting things, because it lets you supply a default (`0`) for keys seen for the first time, avoiding a separate `if key in d` check:

````python
s = input('Enter a string: ')
d = {}
for simb in s:
 d[simb] = d.get(simb, 0) + 1
    
print(d)
````

`dict.setdefault(key[, x])` - returns the value indicated by the key `key`. If there is no such key in the dictionary, creates a new pair `key: x`. If the value `x` is not specified, creates the pair `key: None`.

````python
empl = {'salesman': 'Johnson'}

# the already existing value is not changed, even if another value is specified
print(empl.setdefault('salesman', 'Smith'))

empl.setdefault('age', 30)
empl.setdefault('salary')

print(empl)
````

`dict.pop(key[, x])` - removes a dictionary element and returns the value of `dict[key]`, if such a key exists. If there is no such key, returns `x` or outputs an error message if the value `x` is not specified.

````python
empl = {'salesman': 'Johnson', 'age': 30, 'salary': 1500}

print(empl.pop('salary'))
print(empl.pop('name','There is no such key in the dictionary'))
print(empl)
````

````python
empl = {'salesman': 'Johnson', 'age': 30, 'salary': 1500}

print(empl.pop('name'))
print(empl)
````

````python
d = {'name': 'John', 'age':22, 'salary': 3000, 'city': 'New York'}

d['location'] = d.pop('city') # replaces the key 'city' with the key 'location'
print(d)
````

`dict.popitem()` - removes and returns the last dictionary pair, if the dictionary is empty, outputs an error message.

````python
empl = {'salesman': 'Johnson', 'salary': 1500, 'age': 30}

print(empl.popitem())
print(empl)
````

`dict.clear()` - clears the dictionary.

````python
empl = {'salesman': 'Johnson', 'salary': 1500, 'age': 30}

empl.clear()
print(empl)
````

`dict.update(dict1)` - updates the dictionary with the pairs of the dictionary `dict1`. If the keys of the dictionary `dict1` already exist in the dictionary `dict`, replaces the values indicated by the key.

````python
empl = {'job':'salesman', 'salary': 1500, 'lname': 'Johnson' }
print(empl)

empl.update({'fname':'John', 'salary':1700})
print(empl)
````

## Concatenation of sequences

### Concatenation of lists

The `*` operator collects the remaining elements into a list when assigning values to variables. It allows one variable to receive multiple remaining values.

````python
a, b, *c = 1, 2, 3, 4, 5, 6

print(a)  # 1
print(b)  # 2
print(c)  # [3, 4, 5, 6]
````

The `*` operator also unpacks elements from a list into individual elements. It allows us to insert all elements of a list into another list without using the `+` operator.

````python
L1 = ['a','b','c']
L2 = [1, 2, 3]

L3 = L1 + L2
L4 = [*L1, *L2]

print(L3)
print(L4)
````

### Concatenation of tuples

````python
t1 = ('a','b','c', [100,200])
t2 = (1, 2, 3)

t3 = t1 + t2
t4 = (*t1, *t2)

print(t3)
print(t4)
````

### Concatenation of sets

````python
A = {'abc','bcd'}
B = {1, 2, 3, 'abc'}

A_B = {*A, *B}

print(A_B)
````

### Concatenation of dictionaries

````python
d1 = {'one': 100, 'two': 200, 'three': 300}
d2 = {'A': 1, 'B': 2, 'C': 3, 'D': 4, 'one': 5000}

# Method 1: Using the ** operator
d = {**d1, **d2}
print(d)

# Method 2: Using copy() and update()
d = d1.copy()
d.update(d2)
print(d)
````
